# Sunburst Chart Examples

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Sales Performance by Quarter
Source: <https://www.canvasxpress.org/examples/sunburst-1.html>
A circular sunburst chart displaying sales performance, color-coded by quarter. Each quarter is represented by a distinct color, with sales data segmented by month and week, though weekly data with null values is not displayed.

```python
data = {
  'x': {
    'Color': ['red', 'blue', 'green', 'grey', 'red', 'blue', 'green', 'grey', 'red', 'blue', 'green', 'grey', 'red', 'blue', 'green'],
    'Month': ['Jan', 'Feb', 'Feb', 'Feb', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Quarter': ['1st', '1st', '1st', '1st', '1st', '1st', '2nd', '2nd', '2nd', '3rd', '3rd', '3rd', '4th', '4th', '4th'],
    'Week': [None, 'Week 1', 'Week 2', 'Week 3', 'Week 4', None, None, None, None, None, None, None, None, None, None],
  },
  'y': {
    'data': [
      [3.5, 1.2, 0.8, 0.6, 0.5, 1.7, 1.1, 0.8, 0.3, 0.7, 0.6, 0.1, 0.5, 0.4, 0.3],
    ],
    'smps': ['Sales1', 'Sales2', 'Sales3', 'Sales4', 'Sales5', 'Sales6', 'Sales7', 'Sales8', 'Sales9', 'Sales10', 'Sales11', 'Sales12', 'Sales13', 'Sales14', 'Sales15'],
    'vars': ['Sales'],
  },
}
config = {
  'circularArc': 360,
  'circularRotate': 0,
  'circularType': 'sunburst',
  'colorBy': 'Quarter',
  'colorScheme': 'Bootstrap',
  'graphType': 'Circular',
  'hierarchy': ['Quarter', 'Month', 'Week'],
  'objectBorderColor': 'rgb(0,0,0)',
  'showTransition': False,
  'title': 'Simple Sunburst',
  'xAxis': ['Sales'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

The chart should display sales data with a circular arc of 360 degrees, showing monthly and weekly breakdowns, but omitting weeks with null values. Focus on quarterly performance with specific color coding for each quarter.
