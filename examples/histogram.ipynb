## Example 4: Scatter Plot of Tobacco vs. Alcohol Spending by Region
Source: <https://www.canvasxpress.org/examples/histogram-4.html>
A scatter plot with histograms showing weekly household spending on tobacco products and alcoholic beverages across 11 regions of Great Britain.

```python
data = {
  'y': {
    'data': [
      [6.47, 4.03],
      [6.13, 3.76],
      [6.19, 3.77],
      [4.89, 3.34],
      [5.63, 3.47],
      [4.52, 2.92],
      [5.89, 3.2],
      [4.79, 2.71],
      [3.53, 3.53],
      [6.08, 4.51],
      [4.02, 4.56],
    ],
    'smps': ['Alcohol', 'Tobacco'],
    'vars': ['North', 'Yorkshire', 'Northeast', 'East Midlands', 'West Midlands', 'East Anglia', 'Southeast', 'Southwest', 'Wales', 'Scotland', 'Northern Ireland'],
  },
}
config = {
  'citation': 'Moore, David S., and George P. McCabe (1989). Introduction to the Practice of Statistics, p. 179.',
  'graphType': 'Scatter2D',
  'histogramBins': 5,
  'showHistogram': True,
  'title': 'Average weekly household spending, in British pounds, on tobacco products\\nand alcoholic beverages for each of the 11 regions of Great Britain.',
  'xAxis': ['Tobacco', 'Alcohol'],
  'xAxisTitle': 'Pounds Spent',
  'yAxisTitle': 'Frequency',
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** To generate this chart, please use CanvasXpress and provide the following details: 1) Scatter2D chart type, 2) X-axis as 'Tobacco' and 'Alcohol', 3) Y-axis showing frequency, 4) Include histograms for data distribution, 5) Use regions as data points with specific spending values, 6) Highlight key spending patterns between tobacco and alcohol.

## Example 5: Comparison of Average Weekly Household Spending on Tobacco and Alcohol Across Regions of Great Britain
Source: <https://www.canvasxpress.org/examples/histogram-5.html>
A scatter plot comparing average weekly household spending on tobacco products and alcoholic beverages across 11 regions in Great Britain. The x-axis represents pounds spent on tobacco and alcohol, while the y-axis shows frequency. The data is sourced from Moore, David S., and George P. McCabe (1989). Introduction to the Practice of Statistics, p. 179.

```python
data = {
  'y': {
    'data': [
      [6.47, 4.03],
      [6.13, 3.76],
      [6.19, 3.77],
      [4.89, 3.34],
      [5.63, 3.47],
      [4.52, 2.92],
      [5.89, 3.2],
      [4.79, 2.71],
      [3.53, 3.53],
      [6.08, 4.51],
      [4.02, 4.56],
    ],
    'smps': ['Alcohol', 'Tobacco'],
    'vars': ['North', 'Yorkshire', 'Northeast', 'East Midlands', 'West Midlands', 'East Anglia', 'Southeast', 'Southwest', 'Wales', 'Scotland', 'Northern Ireland'],
  },
}
config = {
  'citation': 'Moore, David S., and George P. McCabe (1989). Introduction to the Practice of Statistics, p. 179.',
  'graphType': 'Scatter2D',
  'histogramBins': 5,
  'histogramType': 'staggered',
  'showHistogram': True,
  'title': 'Average weekly household spending, in British pounds, on tobacco products\\nand alcoholic beverages for each of the 11 regions of Great Britain.',
  'xAxis': ['Tobacco', 'Alcohol'],
  'xAxisTitle': 'Pounds Spent',
  'yAxisTitle': 'Frequency',
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** Create a scatter plot using CanvasXpress to visualize the comparison of average weekly household spending on tobacco products and alcoholic beverages across 11 regions in Great Britain. The x-axis represents pounds spent on tobacco, and the y-axis represents pounds spent on alcohol. Use frequency as the data point. Highlight the two variables on the x-axis and the 11 regions on the y-axis. Mention the staggered histograms and the reference to Moore and McCabe's book. Analyze the distribution, correlation, and notable patterns between the two products.
