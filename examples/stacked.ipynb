## Example 2: Horizontal Stacked Bar Chart with Multiple Factors
Source: <https://www.canvasxpress.org/examples/stacked-2.html>
A horizontal stacked bar chart displaying data across multiple factors. The x-axis shows Factor1 to Factor4, while the y-axis shows Factor5 to Factor6. The chart illustrates the magnitude of each sample (S1 to S6) across different variables.

```python
data = {
  'x': {
    'Factor1': ['Lev : 1', 'Lev : 2', 'Lev : 3', 'Lev : 1', 'Lev : 2', 'Lev : 3'],
    'Factor2': ['Lev : A', 'Lev : B', 'Lev : A', 'Lev : B', 'Lev : A', 'Lev : B'],
    'Factor3': ['Lev : X', 'Lev : X', 'Lev : Y', 'Lev : Y', 'Lev : Z', 'Lev : Z'],
    'Factor4': [5, 10, 15, 20, 25, 30],
    'Factor5': [8, 16, 24, 32, 40, 48],
    'Factor6': [10, 20, 30, 40, 50, 60],
  },
  'y': {
    'data': [
      [5, 10, 25, 40, 45, 50],
      [95, 80, 75, 70, 55, 40],
      [25, 30, 45, 60, 65, 70],
      [55, 40, 35, 30, 15, 1],
    ],
    'desc': ['Magnitude1', 'Magnitude2'],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4'],
  },
  'z': {
    'Annt1': ['Desc : 1', 'Desc : 2', 'Desc : 3', 'Desc : 4'],
    'Annt2': ['Desc : A', 'Desc : B', 'Desc : A', 'Desc : B'],
    'Annt3': ['Desc : X', 'Desc : X', 'Desc : Y', 'Desc : Y'],
    'Annt4': [5, 10, 15, 20],
    'Annt5': [8, 16, 24, 32],
    'Annt6': [10, 20, 30, 40],
  },
}
config = {
  'graphOrientation': 'horizontal',
  'graphType': 'Stacked',
  'legendBackgroundColor': False,
  'sampleSpaceFactor': 1,
  'showDataValues': True,
  'smpTextScaleFontFactor': 0.8,
  'smpTitle': 'Collection of Samples',
  'smpTitleFontStyle': 'italic',
  'title': 'Random Data',
  'xAxis': ['V1', 'V2', 'V3', 'V4'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 5: Stacked Horizontal Analysis of Three Characters' Truthfulness
Source: <https://www.canvasxpress.org/examples/stacked-5.html>
A horizontal stacked graph comparing the truthfulness levels across three characters, showing the distribution of various truthfulness variables.

```python
data = {
  'y': {
    'data': [
      [-35, -20, -5],
      [-25, -20, -10],
      [-15, -10, -10],
      [10, 10, 15],
      [10, 20, 25],
      [5, 20, 35],
    ],
    'smps': ['Big Fat Liar', 'Two-Faced', 'Honest Abe'],
    'vars': ['Pants on Fire', 'False', 'Mostly False', 'Half True', 'Mostly True', 'True'],
  },
}
config = {
  'axisAlgorithm': 'wilkinson',
  'colorScheme': 'CanvasXpress',
  'graphOrientation': 'horizontal',
  'graphType': 'Stacked',
  'legendColumns': 3,
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'legendPosition': 'bottom',
  'marginRight': 20,
  'showDataValues': True,
  'title': 'Diverging Stacked Graph',
  'xAxisTickFormat': '%s%%',
  'xAxis': ['Pants on Fire', 'False', 'Mostly False', 'Half True', 'Mostly True', 'True'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```
