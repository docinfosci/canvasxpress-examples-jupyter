## Example 1: Correlation Analysis of Variables with Samples
Source: <https://www.canvasxpress.org/examples/correlation-1.html>
A correlation plot showing the relationship between variables V1 to V4 across different samples. The x-axis represents the variables, while the y-axis shows the correlation values. The data includes multiple samples and variables, with some notable patterns in the distribution of correlation values.

```python
data = {
  'x': {
    'Factor1': ['Lev : 1', 'Lev : 2', 'Lev : 3', 'Lev : 1', 'Lev : 2', 'Lev : 3'],
    'Factor2': ['Lev : A', 'Lev : B', 'Lev : A', 'Lev : B', 'Lev : A', 'Lev : B'],
    'Factor3': ['Lev : X', 'Lev : X', 'Lev : Y', 'Lev : Y', 'Lev : Z', 'Lev : Z'],
    'Factor4': [5, 10, 15, 20, 25, 30],
    'Factor5': [8, 16, 24, 32, 40, 48],
    'Factor6': [10, 20, 30, 40, 50, 60],
  },
  'y': {
    'data': [
      [5, 10, 25, 40, 45, 50],
      [95, 80, 75, 70, 55, 40],
      [25, 30, 45, 60, 65, 70],
      [55, 40, 35, 30, 15, 1],
    ],
    'desc': ['Magnitude1', 'Magnitude2'],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4'],
  },
  'z': {
    'Annt1': ['Desc : 1', 'Desc : 2', 'Desc : 3', 'Desc : 4'],
    'Annt2': ['Desc : A', 'Desc : B', 'Desc : A', 'Desc : B'],
    'Annt3': ['Desc : X', 'Desc : X', 'Desc : Y', 'Desc : Y'],
    'Annt4': [5, 10, 15, 20],
    'Annt5': [8, 16, 24, 32],
    'Annt6': [10, 20, 30, 40],
  },
}
config = {
  'correlationAxis': 'samples',
  'graphType': 'Correlation',
  'title': 'Correlation Plot',
  'yAxisTitle': 'Correlation Title',
  'xAxis': ['V1', 'V2', 'V3', 'V4'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 3: Variable Correlations Across Magnitudes
Source: <https://www.canvasxpress.org/examples/correlation-3.html>
A correlation plot displaying relationships between numerical variables V1 to V4 across two magnitude categories. The x-axis represents the variables, while the y-axis shows the magnitude categories. The plot highlights correlations and patterns in the data.

```python
data = {
  'x': {
    'Factor1': ['Lev : 1', 'Lev : 2', 'Lev : 3', 'Lev : 1', 'Lev : 2', 'Lev : 3'],
    'Factor2': ['Lev : A', 'Lev : B', 'Lev : A', 'Lev : B', 'Lev : A', 'Lev : B'],
    'Factor3': ['Lev : X', 'Lev : X', 'Lev : Y', 'Lev : Y', 'Lev : Z', 'Lev : Z'],
    'Factor4': [5, 10, 15, 20, 25, 30],
    'Factor5': [8, 16, 24, 32, 40, 48],
    'Factor6': [10, 20, 30, 40, 50, 60],
  },
  'y': {
    'data': [
      [5, 10, 25, 40, 45, 50],
      [95, 80, 75, 70, 55, 40],
      [25, 30, 45, 60, 65, 70],
      [55, 40, 35, 30, 15, 1],
    ],
    'desc': ['Magnitude1', 'Magnitude2'],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4'],
  },
  'z': {
    'Annt1': ['Desc : 1', 'Desc : 2', 'Desc : 3', 'Desc : 4'],
    'Annt2': ['Desc : A', 'Desc : B', 'Desc : A', 'Desc : B'],
    'Annt3': ['Desc : X', 'Desc : X', 'Desc : Y', 'Desc : Y'],
    'Annt4': [5, 10, 15, 20],
    'Annt5': [8, 16, 24, 32],
    'Annt6': [10, 20, 30, 40],
  },
}
config = {
  'correlationAnchorLegend': True,
  'correlationAnchorLegendAlignWidth': 20,
  'correlationAxis': 'variables',
  'graphType': 'Correlation',
  'title': 'Correlation Plot',
  'yAxisTitle': 'Correlation Title',
  'xAxis': ['V1', 'V2', 'V3', 'V4'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```
