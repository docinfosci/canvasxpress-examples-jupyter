# Bullet Chart Examples

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Horizontal Bullet Chart Comparing Variables V1 to V5 Across Samples S1 to S6
Source: <https://www.canvasxpress.org/examples/bullet-1.html>
A horizontal bullet chart comparing variables V1-V5 across samples S1-S6. Each sample is represented by a bullet point for each variable. V1 values are consistently higher than other variables, and V5 shows the highest values among S3, S4, and S6.

```python
data = {
  'y': {
    'data': [
      [19, 25, 52, 57, 31, 42],
      [15, 15, 15, 20, 20, 20],
      [35, 35, 35, 40, 40, 40],
      [70, 70, 70, 70, 70, 70],
      [46, 52, 42, 63, 56, 51],
    ],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4', 'V5'],
  },
}
config = {
  'rangeColors': ['#777777', '#AAAAAA', '#DDDDDD'],
  'rangeStack': ['V2', 'V3', 'V4'],
  'bulletTargetVarName': 'V5',
  'dataTextColor': '#000000',
  'dataValuesPosition': 'inside',
  'graphOrientation': 'horizontal',
  'graphType': 'Bullet',
  'showDataValues': True,
  'showLegend': False,
  'xAxis': ['V1'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** I need to create a horizontal bullet chart showing the comparison of variables V1 to V5 across samples S1 to S6. The chart should use a spectrum of colors from dark gray to medium gray. V1 is on the x-axis, and the variables V2 to V5 should be in the legend. The bullets should represent the data values, with V1 having the highest values and V5 showing the highest values in S3, S4, and S6. Please include the key data points and notable patterns in the chart.

## Example 2: Comparison of Sample Values Across Variables
Source: <https://www.canvasxpress.org/examples/bullet-2.html>
A vertical bullet chart comparing sample values (S1-S6) across variables V1-V5. V1 is on the x-axis, with V2, V3, and V4 stacked in a range plot. The chart uses a range color scheme for V2, V3, and V4, and highlights V5 with a distinct color. Notable patterns include higher values in V1 and V5 for samples S4 and S6.

```python
data = {
  'y': {
    'data': [
      [19, 25, 52, 57, 31, 42],
      [15, 15, 15, 20, 20, 20],
      [35, 35, 35, 40, 40, 40],
      [70, 70, 70, 70, 70, 70],
      [46, 52, 42, 63, 56, 51],
    ],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4', 'V5'],
  },
}
config = {
  'colors': ['#3F3F3F'],
  'graphOrientation': 'vertical',
  'graphType': 'Bullet',
  'showDataValues': True,
  'smpTextRotate': '90',
  'xAxis': ['V1'],
  'rangeStack': ['V2', 'V3', 'V4'],
  'rangeColors': ['#945D55', '#C4A285', '#EBE7DE'],
  'bulletTargetVarName': 'V5',
  'bulletTargetVarColor': '#3F3F3F',
  'segregateSamplesBy': ['sample'],
  'layoutTopology': '1X6',
  'layoutAdjust': True,
  'stripShow': False,
  'xAxisTextScaleFontFactor': 0.6,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** Could you generate a vertical bullet chart using CanvasXpress that compares sample values (S1-S6) across variables V1-V5? The x-axis should represent V1, with V2, V3, and V4 stacked in a range plot. Use the following colors: #945D55 for V2, #C4A285 for V3, and #EBE7DE for V4. Highlight V5 with the color #3F3F3F. Include a legend, show data values, rotate sample text by 90 degrees, and use a 1x6 layout. Look for trends in sample values across variables.

## Example 3: Color vs. Sample
Source: <https://www.canvasxpress.org/examples/bullet-3.html>
A vertical bullet chart showing the relationship between color categories (A, B, C) and sample points, with each category's data point displayed at different heights.

```python
data = {
  'x': {
    'Color': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
  },
  'y': {
    'data': [
      [10, 12, 22, 31, 42, 27, 15, 15, 5],
    ],
    'smps': ['Birth', 'Puberty', 'Death', 'Family', 'Friendship', 'Love', 'Work', 'Passion', 'Environment'],
    'vars': ['Topic'],
  },
}
config = {
  'backgroundType': 'solid',
  'rangeColors': ['rgb(48,126,164)'],
  'rangeColorTransparency': 1,
  'bulletWidthRatio': 1,
  'colorBy': 'Color',
  'colors': ['rgb(250,165,44)', 'rgb(254,225,60)', 'rgb(253,243,169)'],
  'fontName': 'Waltograph',
  'fontsExternal': [
    {
      'name': 'Waltograph',
      'url': 'https://www.canvasxpress.org/assets/fonts/waltograph42.otf',
    },
  ],
  'graphType': 'Bullet',
  'graphOrientation': 'vertical',
  'marginBottom': 0,
  'marginLeft': 50,
  'marginRight': 50,
  'marginTop': 50,
  'maxTextSize': 80,
  'objectBorderColor': 'rgba(255,255,255,0)',
  'plotBackgroundColor': 'rgb(48,114,148)',
  'showLegend': False,
  'smpTextColor': 'rgb(255,255,255)',
  'smpTextRotate': '30',
  'smpTextScaleFontFactor': 2,
  'title': 'Friendship, Love, Family',
  'titleAlign': 'center',
  'titleColor': 'rgb(254,225,60)',
  'titleScaleFontFactor': 2.5,
  'xAxis': ['Topic'],
  'xAxisShow': False,
  'xAxis2Show': False,
  'xAxisGridMajorShow': False,
  'xAxisGridMinorShow': False,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** To create this chart using CanvasXpress, you would start by selecting the 'Bullet' chart type. Set the y-axis to display the data points for each sample, with categories A, B, and C represented by different colors. Use the specified range colors and ensure the font style is 'Waltograph'. The chart should be vertical, with each category's data point clearly visible at its respective height, and the color scheme should differentiate between the three categories smoothly.

## Example 4: Horizontal Stacked Bullet Chart for Countries

## Example 5: Performance Metrics Across Samples

```python
data = {
  'y': {
    'data': [
      [250.43, 5.69, 9.79],
      [250, 3, 8],
      [400, 6, 10],
      [600, 12, 11],
      [520, 7.5, 10.5],
    ],
    'smps': ['Total Distance', 'Average Distance', 'Average Speed'],
    'vars': ['Value', 'Bad', 'Acceptable', 'Good', 'Target'],
  },
}
config = {
  'barType': 'bullet',
  'rangeColors': ['#FFFFFF', '#F2F2F2', '#D8D8D8'],
  'rangeStack': ['Bad', 'Acceptable', 'Good'],
  'rangeStackShow': True,
  'bulletTargetType': 'openCircle',
  'bulletTargetVarName': 'Target',
  'bulletTargetWidthRatio': 0.2,
  'bulletStyle': 'roundGradient',
  'bulletWidthRatio': 0.5,
  'colors': ['#9422F5'],
  'dataValuesPosition': 'inside',
  'dataTextColor': '#FFFFFF',
  'dataTextScaleFontFactor': 0.8,
  'graphOrientation': 'horizontal',
  'graphType': 'Bullet',
  'layoutAdjust': True,
  'layoutSpacing': 20,
  'layoutTopology': '3X1',
  'segregateSamplesBy': ['sample'],
  'showDataValues': True,
  'showLegend': False,
  'showSampleNames': False,
  'stripBackgroundColor': '#FFFFFF',
  'stripTextAlign': 'left',
  'stripTextColor': '#000000',
  'xAxis': ['Value'],
  'xAxis2Show': True,
  'xAxisGridMajorShow': False,
  'xAxisGridMinorShow': False,
  'xAxisShow': False,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 6: NGS Data Progress
Source: <https://www.canvasxpress.org/examples/bullet-6.html>
A horizontal bullet chart showing progress across different NGS data types, with each type represented by a colored progress bar.

```python
data = {
  'y': {
    'vars': ['Samples'],
    'smps': ['RNAseq (all types)', 'WGS', 'ctDNA / TSO500', 'WES / Other', 'GDA + Neuroboost', 'CHIP Mutation', 'Other'],
    'data': [
      [26, 16, 6, 1, 3, 2, 5],
    ],
  },
  'x': {
    'Type': ['RNAseq', 'WGS', 'ctDNA', 'WES', 'GDA', 'CHIP', 'Other'],
  },
}
config = {
  'graphType': 'Bullet',
  'bulletStyle': 'progress',
  'graphOrientation': 'horizontal',
  'progressWidthRatio': 0.28,
  'colorBy': 'Type',
  'colors': ['rgb(200,150,225)', 'rgb(240,170,90)', 'rgb(150,160,175)', 'rgb(80,160,235)', 'rgb(150,160,180)', 'rgb(90,200,120)', 'rgb(240,120,105)'],
  'backgroundType': 'solid',
  'background': 'rgb(20,26,38)',
  'plotBackgroundColor': 'rgb(20,26,38)',
  'foreground': 'rgb(235,240,248)',
  'objectBorderColor': 'rgba(0,0,0,0)',
  'showDataValues': True,
  'showLegend': False,
  'dataTextFontStyle': 'bold',
  'smpTextColor': 'rgb(235,240,248)',
  'title': 'NGS Data Types',
  'titleAlign': 'left',
  'titleColor': 'rgb(150,160,180)',
  'titleScaleFontFactor': 1.4,
  'xAxis': ['Samples'],
  'xAxisShow': False,
  'xAxis2Show': False,
  'xAxisGridMajorShow': False,
  'xAxisGridMinorShow': False,
  'yAxisGridMajorShow': False,
  'marginTop': 40,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** I need a horizontal bullet chart showing NGS data progress. The chart should display progress bars for RNAseq, WGS, ctDNA, WES, GDA, CHIP, and Other types. The chart should use specific colors corresponding to each type and show progress towards a goal. The chart should have a legend indicating the data types and progress thresholds. Key parameters include the total samples and progress percentage. The chart should highlight key data points such as 26% for RNAseq, 16% for WGS, and so on. The title should reflect the overall data progress, and the description should explain the chart's purpose and data structure.
