# Lollipop Chart Examples

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Distribution of V1 Values Across Samples
Source: <https://www.canvasxpress.org/examples/lollipop-1.html>
A bar chart showing the distribution of V1 values across different samples (S1 to S8). The chart uses a lollipop style with data points colored according to the CanvasXpress color scheme. The x-axis represents the samples, and the y-axis shows the range of V1 values.

```python
data = {
  'y': {
    'data': [
      [10, 15, 20, 30, 40, 70, 80, 90],
    ],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6', 'S7', 'S8'],
    'vars': ['V1'],
  },
  'x': {
    'val': [3, 3, 3, 3, 5, 3, 3, 1],
  },
}
config = {
  'barType': 'lollipop',
  'colorScheme': 'CanvasXpress',
  'dataPointSizeScaleFactor': 6,
  'graphType': 'Bar',
  'widthFactor': 0.2,
  'sizeBy': 'val',
  'xAxis': ['V1'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 2: Performance by Category
Source: <https://www.canvasxpress.org/examples/lollipop-2.html>
A bar chart showing performance values for different categories. The x-axis represents categories labeled A to F, and the y-axis shows the performance metric Var1 with values ranging from -85 to 100.

```python
data = {
  'x': {
    'Color': ['A', 'B', 'C', 'D', 'E', 'F'],
  },
  'y': {
    'data': [
      [-40, -85, -60, 50, 75, 100],
    ],
    'smps': ['Engineering', 'Biology', 'Computer', 'Psycology', 'Education', 'Health'],
    'vars': ['Var1'],
  },
}
config = {
  'barLollipopOpen': True,
  'barType': 'lollipopBullet',
  'colorBy': 'Color',
  'colorScheme': 'GGPlot',
  'dataPointSizeScaleFactor': 7,
  'graphType': 'Bar',
  'marginBottom': 50,
  'marginLeft': 50,
  'marginRight': 50,
  'marginTop': 50,
  'maxTextSize': 80,
  'rangeColors': ['rgb(200,200,200)'],
  'setMaxX': 150,
  'setMinX': -150,
  'showDataValues': True,
  'showLegend': False,
  'title': 'Occupations',
  'xAxis2Show': False,
  'xAxisGridMajorShow': False,
  'xAxisGridMinorShow': False,
  'xAxisShow': False,
  'xAxis': ['Var1'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

The chart should use color schemes like GGPlot, show data points with a size scaling factor of 7, and include a lollipop bullet type for the bars. The chart should highlight notable patterns such as variations in performance across categories.
