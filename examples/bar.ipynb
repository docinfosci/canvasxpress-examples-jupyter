# Bar Chart Examples

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Single Variable Bar Chart Showing Category Comparisons
Source: <https://www.canvasxpress.org/examples/bar-1.html>
A vertical bar chart displaying the values of Var 1 across five categories. The categories are labeled 'Cat 1' to 'Cat 5' and are rotated 90 degrees for better readability.

```python
data = {
  'y': {
    'data': [
      [4, 5, 4, 4, 7],
    ],
    'smps': ['Cat 1', 'Cat 2', 'Cat 3', 'Cat 4', 'Cat 5'],
    'vars': ['Var 1'],
  },
}
config = {
  'graphOrientation': 'vertical',
  'graphType': 'Bar',
  'showLegend': False,
  'smpTextRotate': 90,
  'smpTitle': 'Categories',
  'title': 'Bar graph with a single series',
  'xAxis': ['Var 1'],
  'xAxisTitle': 'Var 1',
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

The chart title should reflect the single variable being measured, and no legend should be shown as there is only one series.
