## Example 1: Monthly Revenue Trend
Source: <https://www.canvasxpress.org/examples/area-1.html>

```python
data = {
  'y': {
    'data': [
      [10, 11, 13, 4, 18, 21],
    ],
    'smps': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'vars': ['Value'],
  },
}
config = {
  'colorScheme': 'Prism',
  'graphOrientation': 'vertical',
  'graphType': 'Area',
  'showLegend': False,
  'smpTextRotate': 90,
  'smpTitle': 'Month / First Quarters of 2024',
  'title': 'Area graph with one series',
  'titleScaleFontFactor': 1.2,
  'xAxis': ['Value'],
  'xAxisTitle': 'Revenue (in Millions)',
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 3: Three Series Area Chart Showing Monthly Data

## Example 6: Revenue Trends Across Months with Three Series Overlap
Source: <https://www.canvasxpress.org/examples/area-6.html>
An area chart displaying three revenue series (Series A, B, C) over six months. The chart highlights the overlapping trends of each series, with Series C marked in a distinct color. The y-axis represents revenue in millions.

```python
data = {
  'y': {
    'data': [
      [10, 11, 13, 4, 18, 21],
      [5, 6, 10, 12, 15, 10],
      [2, 3, 6, 4, 8, 6],
    ],
    'smps': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'vars': ['Series A', 'Series B', 'Series C'],
  },
}
config = {
  'colorScheme': 'Behance',
  'filterData': [
    [
      'guess',
      False,
      'different',
      ['Series C'],
    ],
  ],
  'graphOrientation': 'vertical',
  'graphType': 'Area',
  'lineType': 'spline',
  'objectColorTransparency': 0.7,
  'smpTextRotate': 90,
  'smpTitle': 'Month / First Quarters of 2024',
  'subtitle': 'Filtered data',
  'title': 'Area graph with three overlapping data series',
  'titleScaleFontFactor': 1.2,
  'xAxis': ['Series A', 'Series B', 'Series C'],
  'xAxisTitle': 'Revenue (in Millions)',
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 7: Stacked Area Chart Showing Three Series Over Months

## Example 8: Monthly Value Trends Across Three Series
