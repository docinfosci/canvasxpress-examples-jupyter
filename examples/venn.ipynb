# Venn Chart Examples

**[Launch in MyBinder](https://mybinder.org/v2/gh/docinfosci/canvasxpress-examples-jupyter/main?urlpath=lab/tree/examples/venn.ipynb)** for live interactive exploration

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Interactive Venn Diagram Showing Overlaps Between Four Lists
Source: <https://www.canvasxpress.org/examples/venn-1.html>
A Venn diagram illustrating the intersections and unique elements of four lists. The chart displays the size of each intersection, with List 1 (A) having 340 unique elements, List 2 (B) with 562, List 3 (C) with 620, and List 4 (D) with 592. The overlapping regions indicate shared elements between the lists.

```python
data = {
  'venn': {
    'data': {
      'A': 340,
      'AB': 639,
      'ABC': 552,
      'ABCD': 148,
      'ABD': 578,
      'AC': 456,
      'ACD': 298,
      'AD': 257,
      'B': 562,
      'BC': 915,
      'BCD': 613,
      'BD': 354,
      'C': 620,
      'CD': 143,
      'D': 592,
    },
    'legend': {
      'A': 'List 1',
      'B': 'List 2',
      'C': 'List 3',
      'D': 'List 4',
    },
  },
}
config = {
  'graphType': 'Venn',
  'showTransition': False,
  'vennGroups': 4,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** To create this Venn diagram in CanvasXpress, use the Venn diagram chart type with 4 vennGroups. The data represents the size of each intersection among four lists: List 1 (A), List 2 (B), List 3 (C), and List 4 (D). The interactive chart shows the number of elements in each unique region and overlap. The notable large overlap between List 2 (B) and List 3 (C) at 915 elements stands out. Additionally, the significant size of List 1 (A) at 340 elements and List 2 (B) at 562 elements should be highlighted. Use the legend to identify each list and its corresponding color scheme.

## Example 2: Four-Set Venn Diagram Showing Overlaps Between Four Lists
Source: <https://www.canvasxpress.org/examples/venn-2.html>
A Venn diagram illustrating the overlaps between four distinct lists. The diagram shows the intersections and unique elements of List 1, List 2, List 3, and List 4. The largest intersections are ABC (456), BC (915), and BCD (613). The smallest intersection is ABCD (148). The unique elements in each list are also displayed.

```python
data = {
  'venn': {
    'data': {
      'A': 340,
      'AB': 639,
      'ABC': 552,
      'ABCD': 148,
      'ABD': 578,
      'AC': 456,
      'ACD': 298,
      'AD': 257,
      'B': 562,
      'BC': 915,
      'BCD': 613,
      'BD': 354,
      'C': 620,
      'CD': 143,
      'D': 592,
    },
    'legend': {
      'A': 'List 1',
      'B': 'List 2',
      'C': 'List 3',
      'D': 'List 4',
    },
  },
}
config = {
  'graphType': 'Venn',
  'showTransition': False,
  'vennGroups': 3,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```
