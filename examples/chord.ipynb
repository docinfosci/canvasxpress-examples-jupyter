# Chord Chart Examples

**[Launch in MyBinder](https://mybinder.org/docinfosci/canvasxpress-examples-jupyter/main?urlpath=lab/tree/examples/chord.ipynb)** for live interactive exploration

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph
```

## Example 1: Simple Chord Graph
Source: <https://www.canvasxpress.org/examples/chord-1.html>
A circular chord graph displaying data values between four categories (Smps A-D) and four variables (Vars A-D). Each chord represents the relationship between an Smp and a Var, with the thickness indicating the magnitude of the value.

```python
data = {
  'y': {
    'data': [
      [11975, 5871, 8916, 2868],
      [1951, 10048, 2060, 6171],
      [8010, 16145, 8090, 8045],
      [1013, 990, 940, 6907],
    ],
    'smps': ['A', 'B', 'C', 'D'],
    'vars': ['A', 'B', 'C', 'D'],
  },
}
config = {
  'circularArc': 360,
  'circularRotate': 0,
  'circularType': 'chord',
  'colors': ['#000000', '#FFDD89', '#957244', '#F26223'],
  'graphType': 'Circular',
  'higlightGreyOut': True,
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'objectBorderColor': 'rgb(0,0,0)',
  'rAxisTickFormat': ['%sK', ' / 1000'],
  'showTransition': False,
  'title': 'Simple Chord Graph',
  'transitionStep': 50,
  'transitionTime': 1500,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

The data shows four categories with corresponding values, and each chord will represent the relationship between an Smp and a Var. The graph should use the specified colors, include a legend for clarity, and highlight the key data points. The circular arc is set to 360 degrees, and the y-axis displays values in thousands with a specific format. Please ensure the chart clearly shows the connections and patterns between the categories.

## Example 2: Interactive Rotated Chord Graph for 4x4 Data Matrix
Source: <https://www.canvasxpress.org/examples/chord-2.html>
A circular chord graph displaying a 4x4 data matrix. Each chord connects corresponding values in the matrix, with a 180-degree circular rotation and 360-degree circular arcs. The largest values are highlighted in gray.

```python
data = {
  'y': {
    'data': [
      [11975, 5871, 8916, 2868],
      [1951, 10048, 2060, 6171],
      [8010, 16145, 8090, 8045],
      [1013, 990, 940, 6907],
    ],
    'smps': ['A', 'B', 'C', 'D'],
    'vars': ['A', 'B', 'C', 'D'],
  },
}
config = {
  'chordColor': 'largest',
  'circularArc': 360,
  'circularRotate': 180,
  'circularType': 'chord',
  'graphType': 'Circular',
  'higlightGreyOut': True,
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'objectBorderColor': 'rgb(0,0,0)',
  'rAxisTickFormat': ['%sK', ' / 1000'],
  'showTransition': False,
  'title': 'Rotated Chord Graph',
  'transitionStep': 50,
  'transitionTime': 1500,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 3: Interactive 4x4 Chord Diagram
Source: <https://www.canvasxpress.org/examples/chord-3.html>
An interactive chord diagram showing relationships between four groups (smps) across four variables (vars). Each chord represents the magnitude of the relationship between the respective group and variable.

```python
data = {
  'y': {
    'data': [
      [11975, 5871, 8916, 2868],
      [1951, 10048, 2060, 6171],
      [8010, 16145, 8090, 8045],
      [1013, 990, 940, 6907],
    ],
    'smps': ['A', 'B', 'C', 'D'],
    'vars': ['A', 'B', 'C', 'D'],
  },
}
config = {
  'circularArc': 180,
  'circularRotate': -90,
  'circularType': 'chord',
  'graphType': 'Circular',
  'higlightGreyOut': True,
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'objectBorderColor': 'rgb(0,0,0)',
  'rAxisTickFormat': ['%sK', ' / 1000'],
  'showLegend': False,
  'showTransition': False,
  'title': 'Rotated Half Chord Graph',
  'transitionStep': 50,
  'transitionTime': 1500,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

**Agent Prompt:** To visualize this data in CanvasXpress, select a 4x4 chord diagram with groups labeled A, B, C, D on both axes. Each chord's thickness indicates the magnitude of the relationship between the corresponding group and variable. The chart is rotated 180 degrees and uses a half-circle layout. Use the circular rotate feature to adjust the view. Note the interactive elements such as hover effects and the ability to zoom in on specific chords to explore the data in detail.

## Example 4: Four-way Round Robin Tournament Similarity Matrix
Source: <https://www.canvasxpress.org/examples/chord-4.html>
A circular chord diagram showing the similarity matrix of a four-team round-robin tournament, with each team's connections highlighting their pairwise similarities.
