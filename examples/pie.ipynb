## Example 1: 2x3 Grid of Pie Charts Showing Magnitude Comparison Across Six Samples
Source: <https://www.canvasxpress.org/examples/pie-1.html>
A 2x3 grid of pie charts displaying magnitude 1 and 2 against six samples, with each chart's layout and data precision noted.

```python
data = {
  'x': {
    'Factor1': ['Lev : 1', 'Lev : 2', 'Lev : 3', 'Lev : 1', 'Lev : 2', 'Lev : 3'],
    'Factor2': ['Lev : A', 'Lev : B', 'Lev : A', 'Lev : B', 'Lev : A', 'Lev : B'],
    'Factor3': ['Lev : X', 'Lev : X', 'Lev : Y', 'Lev : Y', 'Lev : Z', 'Lev : Z'],
    'Factor4': [5, 10, 15, 20, 25, 30],
    'Factor5': [8, 16, 24, 32, 40, 48],
    'Factor6': [10, 20, 30, 40, 50, 60],
  },
  'y': {
    'data': [
      [5, 10, 25, 40, 45, 50],
      [95, 80, 75, 70, 55, 40],
      [25, 30, 45, 60, 65, 70],
      [55, 40, 35, 30, 15, 1],
    ],
    'desc': ['Magnitude1', 'Magnitude2'],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4'],
  },
  'z': {
    'Annt1': ['Desc : 1', 'Desc : 2', 'Desc : 3', 'Desc : 4'],
    'Annt2': ['Desc : A', 'Desc : B', 'Desc : A', 'Desc : B'],
    'Annt3': ['Desc : X', 'Desc : X', 'Desc : Y', 'Desc : Y'],
    'Annt4': [5, 10, 15, 20],
    'Annt5': [8, 16, 24, 32],
    'Annt6': [10, 20, 30, 40],
  },
}
config = {
  'graphType': 'Pie',
  'layout': '2X3',
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'pieSegmentLabels': 'inside',
  'pieSegmentPrecision': 0,
  'pieSegmentSeparation': 1,
  'showPieGrid': True,
  'showPieSampleLabel': True,
  'showTransition': False,
  'xAxis': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```

## Example 3: Distribution of Factor1 Across Samples
Source: <https://www.canvasxpress.org/examples/pie-3.html>
A pie chart showing the distribution of Factor1 among four samples (S1-S6). Each pie segment represents the proportion of Factor1 within each sample. The chart uses transparent legend background and displays segment values inside the segments.

```python
data = {
  'x': {
    'Factor1': ['Lev : 1', 'Lev : 2', 'Lev : 3', 'Lev : 1', 'Lev : 2', 'Lev : 3'],
    'Factor2': ['Lev : A', 'Lev : B', 'Lev : A', 'Lev : B', 'Lev : A', 'Lev : B'],
    'Factor3': ['Lev : X', 'Lev : X', 'Lev : Y', 'Lev : Y', 'Lev : Z', 'Lev : Z'],
    'Factor4': [5, 10, 15, 20, 25, 30],
    'Factor5': [8, 16, 24, 32, 40, 48],
    'Factor6': [10, 20, 30, 40, 50, 60],
  },
  'y': {
    'data': [
      [5, 10, 25, 40, 45, 50],
      [95, 80, 75, 70, 55, 40],
      [25, 30, 45, 60, 65, 70],
      [55, 40, 35, 30, 15, 1],
    ],
    'desc': ['Magnitude1', 'Magnitude2'],
    'smps': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'vars': ['V1', 'V2', 'V3', 'V4'],
  },
  'z': {
    'Annt1': ['Desc : 1', 'Desc : 2', 'Desc : 3', 'Desc : 4'],
    'Annt2': ['Desc : A', 'Desc : B', 'Desc : A', 'Desc : B'],
    'Annt3': ['Desc : X', 'Desc : X', 'Desc : Y', 'Desc : Y'],
    'Annt4': [5, 10, 15, 20],
    'Annt5': [8, 16, 24, 32],
    'Annt6': [10, 20, 30, 40],
  },
}
config = {
  'graphType': 'Pie',
  'legendKeyBackgroundBorderColor': 'rgba(255,255,255,0)',
  'legendKeyBackgroundColor': 'rgba(255,255,255,0)',
  'pieBy': 'Factor1',
  'pieSegmentLabels': 'inside',
  'pieSegmentPrecision': 0,
  'showTransition': False,
}
cx = CanvasXpress(data=data, config=config)
graph(cx)
```
